```python
from osgeo import gdal
from os import listdir
import wradlib as wrl
import pandas as pd
import datetime

file_path = "D:\\Data\\13May\\IMD-B"
files = listdir(file_path)
radar_data = []
IST = datetime.timedelta(hours=5,minutes=30)

for fname in files:
    fcontent = wrl.io.read_iris(file_path+'\\'+fname)
    time = fcontent['data'][1]['ingest_data_hdrs']['DB_DBZ']['sweep_start_time']+IST
    dtime = datetime.datetime.strftime(time,"%Y-%m-%d %H:%M:00")
    dtime = datetime.datetime.strptime(dtime,"%Y-%m-%d %H:%M:00")
    radar_data.append(dtime)

```

```python
file_path = "D:\\Data\\RRData\\0000000IMD20200513_Extrct"
file = open(file_path,'r')
i,j,rainrate=0,0,0
rrdata = []

for line in file:
    items = line.split()
    time = items[3]+' '+items[2]
    dtime = datetime.datetime.strptime(time,"%d.%m.%Y %H:%M:%S")
    if dtime==radar_data[i]:
        i+=1
        j = 10
        rrdata.append(rainrate)
        rainrate = 0
    elif j:
        rainrate += float(items[0])
        j-=1
        
            
file.close()

time = pd.DataFrame({'DateTime':radar_data[:-6],'RainRate':rrdata})
time.to_csv('time.csv')
```
